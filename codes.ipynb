```python
# ==================== LIST SYNTAX CHEATSHEET ====================

# 1. Creating Lists
a = []
a = [10, 20, 30]
a = [1, "Python", 3.5, True]
a = list((10, 20, 30))
a = list(range(5))

# 2. Accessing Items
a[0]          # First item
a[-1]         # Last item
a[1]          # Item at index 1

# 3. Slicing
# a[start:stop]
a[1:4]
a[:3]
a[2:]
a[:]
a[::2]
a[::-1]
a[-3:]

# 4. Changing Items
a[0] = 100
a[1:3] = [200, 300]
a[1:2] = [20, 30, 40]

# 5. Adding Items
a.append(40)
a.insert(1, 15)
a.extend([50, 60])
a += [70, 80]

# 6. Removing Items
a.remove(20)
a.pop()
a.pop(1)
del a[0]
del a[1:3]
a.clear()

# 7. Searching
20 in a
20 not in a
a.append(20)
a.index(20)
a.count(20)

# 8. Sorting and Reversing
a.sort()
a.sort(reverse=True)
a.reverse()
sorted(a)
sorted(a, reverse=True)

# 9. Copying Lists
b = a.copy()
b = a[:]
b = list(a)

# 10. List Length and Calculations
len(a)
min(a)
max(a)
sum(a)

# 11. List Operators
a + b
a * 3
a == b
a != b
a < b
a > b

# 12. Looping Through a List
for item in a:
    print(item)

for i in range(len(a)):
    print(a[i])

for i, item in enumerate(a):
    print(i, item)

# 13. List Comprehension
[x for x in range(10)]
[x for x in range(10) if x % 2 == 0]
[x * 2 for x in a]
[x if x > 0 else 0 for x in a]

# 14. Nested Lists
a = [[1, 2], [3, 4]]
a[0]
a[0][1]
a[1][0]

# 15. Unpacking
x, y, z = [10, 20, 30]
x, *y = [10, 20, 30, 40]
x, *y, z = [10, 20, 30, 40]

# 16. Useful Built-in Functions
a = [10,20,30]
list(a)
len(a)
sum(a)
min(a)
max(a)
sorted(a)
reversed(a)
enumerate(a)
any(a)
all(a)

# 17. Joining and Repeating
a + b
a * 2

# 18. Removing Duplicates
unique = list(set(a))  # Order not guaranteed
unique = list(dict.fromkeys(a))  # Preserves order

# 19. Swapping Items
a[0], a[1] = a[1], a[0]

# 20. Checking Empty List
if not a:
    print("Empty")

# 21. Creating Lists with range()
list(range(5))
list(range(1, 10))
list(range(0, 10, 2))
list(range(10, 0, -1))

# 22. List Methods Returning None
a.append(10)
a.extend([20, 30])
a.insert(0, 5)
a.sort()
a.reverse()
```

**Tuple**

A set is a collection of unique elements in Python.
- Unordered
- Mutable
- Does not allow duplicate values
- Does not support indexing
- Can store different data types
- Elements must be immutable/hashable, such as integers, strings, and tuples containing hashable elements


```python
a = {1,2,3,4,5}
b = {10,20,30,40}
a.add(11)
print(a)
a.update([23,34,45])
print(a)
#remove() -> Error if not present
#discard() -> does not give if not present
a.remove(3)
#a.remove(55)
print(a)
a.discard(55)
print(a)
```

```python
class student:
  school ="CSTU" #class Attribute
  def __init__(self,name,age):
    self.name = name #object Atttribute
    self.age = age
  @classmethod
  def change_school(cls,name):
    cls.school = name
  @staticmethod
  def info():
    print("This is student class")

s1 = student('tomal','20')
s1.gender = 'male'
print(s1.name,s1.age,s1.school,s1.gender)
student.change_school('NSTU')
print(s1.name,s1.age,s1.school,s1.gender)
s2 = student('Tonu','21')
print(s2.name,s2.age,s2.school)
student.info()

```
