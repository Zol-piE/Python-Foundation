**Membership Operator** test membership in a sequence


*   **in** : true if value present
*   **in not** : true if value not present



```python
numbers = [1,2,3,4,5,6]
num = int(input())
print(num in numbers )
```

**Identity Operator** - compare memory location of two object


*   **is** - true id same object
*   **is in**- true if different object




```python
a = [1,2,3]
b= [1,2,3]
c = a
print(a is b)
print(c is a)
print(a is not b)
```
