**Pandas in Python** <br>
> Pandas is a Python library used for *data analysis* and *manipulation*. It works with **tabular data** like Excel spreadsheets and CSV files.



```python
import pandas as pd
print(pd.__version__)
```

 * **Series**: One-dimensional labeled data.
* **DataFrame**: Two-dimensional tabular data with rows and columns.

Series<br>


*   index,label,value.
*   Stores heterogeneous data types.




```python
s = pd.Series([10,20,30,40,50],name='marks')
print(s)
```

```python
s = pd.Series([10,20,30,40,50],name='marks',index=['a','b','c','d','e'])
print(s)
```

```python
arr = [10,20,30]
s = pd.Series(arr,name='marks',index=['a','b','c'])
print(s)
s = pd.Series(5.5,name='marks',index=['a','b','c'],dtype='float')
print(s)
mp = {'a':1,'b':2,'c':3}
s = pd.Series(mp,name='marks')
print(s)
```

**Accessing Series Elements**
* **loc[]** selects by label.
* **iloc[]** selects by integer position.

```python
s = pd.Series([10,20,30,40])
s.index = ['a','b','c','d']
print(s)
#access by label
print(s['a'])
print(s.loc['a'])
print(s.loc[['a','b']])
print(s.loc['a':'c'])
#access by index Position
print(s[0])
print(s.iloc[0])
print(s.iloc[[0,1,2]])
print(s.iloc[0:3])



```

**Series Attributes**

```python
s = pd.Series([10, 20, 30, 40])
s.index       # Index labels
s.values      # Underlying values
s.dtype       # Data type
s.shape       # Shape, e.g. (4,)
s.ndim        # Number of dimensions: 1
s.size        # Number of elements
s.empty       # Whether Series is empty
s.name        # Series name
s.nbytes      # Bytes used by underlying data
```

#**Arithmetic Operations**

```python
s = pd.Series([10, 20, 30])

print(s + 5)
print(s * 2)
print(s - 5)
print(s / 10)
print(s ** 2)
```

**Operations Between Two Series**

```python
x = pd.Series([10, 20, 30],index=['a','b','c'])
y = pd.Series([5, 10, 15],index=['c','a','b'])
print(x + y)
print(x * y)
print(x - y)
print(x / y)
```

```python
x = pd.Series([10, 20, 30],index=['a','b','c'])
y = pd.Series([5, 10, 15],index=['e','f','b'])
print(x+y)
```

**Filtering a Series**

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s > 25)
print(s[s > 25])
#
```

```python
s = pd.Series([10,20,30],name='Marking',index=['a','b','c'])
print(s)
s.name = 'marks'
s.rename(index={'a' : '100'})
print(s)
```

**Statistical Methods**

```python
s = pd.Series([10, 20, 30, 40, 50])

s.sum()          # Sum
s.mean()         # Mean
s.median()       # Median
s.mode()         # Mode
s.min()          # Minimum
s.max()          # Maximum
s.std()          # Standard deviation
s.var()          # Variance
s.count()        # Non-missing count
s.quantile(0.5)  # 50th percentile
s.describe()     # Statistical summary
s.idxmin()       # Index of minimum
s.idxmax()       # Index of maximum
s.cumsum()       # Cumulative sum
s.cumprod()      # Cumulative product
s.cummin()       # Cumulative minimum
s.cummax()       # Cumulative maximum
```

**Adding and Removing Elements**

```python
import numpy as np
rng = np.random.default_rng()
arr = rng.integers(1,5,size=5,endpoint=True)
ind = ['a','b','c','d','e']
s = pd.Series(arr,name='grade',index = ind)
print(s)
s['f'] = 6
print(s)
s = s.drop('f') #drop returns a series
print(s)
s = s.drop(['a','e'])
print(s)
```

**Missing Values**

```python
s = pd.Series([10, np.nan, 30, None, 50])

s.isna()           # Detect missing values
s.isnull()         # Same as isna()
s.notna()          # Detect non-missing values
s.dropna()         # Remove missing values
s.fillna(0)        # Replace missing values with zero
s.ffill()          # Fill using previous valid value
s.bfill()          # Fill using next valid value
s.isna().sum()     # Count missing values
```

```python
s = pd.Series([10, np.nan, 30, None, 50],['a','b','c','d','e'])
print('detect missing values')
print(s.isna())
print(s.isna().sum())
print('detect non-missing values')
print(s.notna())
print(s.notna().sum())
print('drop missing values')
nS = s.dropna();
print(nS)
s.dropna(inplace=True)
print(s)
s = pd.Series([10, np.nan, 30, None, 50],['a','b','c','d','e'])
nS = s.fillna(999)
print(nS)
nS=s.ffill()
print(nS)
nS=s.bfill()
print(nS)
nS = s.replace(np.nan,999)
print(nS)
```

**Unique Values and Frequency**

```python
s = pd.Series(["A", "B", "A", "C", "B", "A"])

s.unique()            # Unique values as an array
s.nunique()           # Number of unique values
s.value_counts()      # Frequency of each value
s.value_counts(normalize=True)  # Relative frequencies
s.duplicated()        # Identify repeated occurrences
s.drop_duplicates()   # Keep unique occurrences
```

```python
import pandas as pd
s = pd.Series(['a','b','a','b','c','c'])
s.name = 'char'
print(s.unique())
print(s.nunique())
print(s.value_counts())
print(s.value_counts(normalize=True))
print(s.duplicated())
print(s.drop_duplicates)
```

**Sorting**

```python
s = pd.Series([11,8,9,3,2,8,9,1],index=['g','h','c','d','e','a','b','f'])
print(s)
print(s.sort_values())
print(s.sort_values(ascending=False))
print(s.sort_index())
print(s.sort_index(ascending=False))
```

**Other Useful Methods**

```python
s.head()             # First 5 elements
s.tail()             # Last 5 elements
s.sample(3)          # Random sample of 3 elements
s.copy()             # Independent copy
s.rename("NewName")  # Rename Series
s.reindex([0, 1, 2]) # Align to a new index
s.reset_index()      # Move index into a column
s.shift(1)           # Shift values down
s.diff()             # Difference from previous value
s.pct_change()       # Percentage change
s.clip(15, 35)       # Limit values to range
s.between(10, 30)    # Check whether values are in range
s.corr(other)        # Correlation with another Series
s.cov(other)         # Covariance with another Series
```
