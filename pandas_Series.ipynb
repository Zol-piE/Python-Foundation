# Pandas DataFrame: Complete Cheat Sheet

## 1. What Is a DataFrame?

A DataFrame is a two-dimensional, labeled data structure in Pandas. It organizes data into rows and columns, similar to an Excel spreadsheet or SQL table.

- Rows represent records.
- Columns represent variables.
- Index labels identify rows.
- Each column is generally a Pandas Series.
- Different columns can have different data types.

```python
import pandas as pd
```

## 2. Creating a DataFrame

### From a Dictionary

```python
data = {
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [21, 22, 20, 23],
    "score": [85, 90, 78, 95]
}

df = pd.DataFrame(data)
print(df)
```

### From a List of Lists

```python
df = pd.DataFrame(
    [[1, "Alice"], [2, "Bob"]],
    columns=["id", "name"]
)
```

### From a List of Dictionaries

```python
df = pd.DataFrame([
    {"name": "Alice", "age": 21},
    {"name": "Bob", "age": 22}
])
```

### With a Custom Index

```python
df = pd.DataFrame(
    {"score": [85, 90, 95]},
    index=["Alice", "Bob", "Charlie"]
)
```

### Create an Empty DataFrame

```python
df = pd.DataFrame()
```

## 5. Selecting Rows with loc and iloc

### loc: Label-Based Selection

`loc` selects rows and columns using their labels.

```python
df.loc[0]                       # Row with index label 0
df.loc[0, "name"]               # One cell
df.loc[0:2]                     # Labels 0, 1, 2
df.loc[:, "name"]               # All rows, name column
df.loc[:, ["name", "score"]]    # Multiple columns
df.loc[0:2, "name":"score"]     # Label-based row and column slices
```

### iloc: Position-Based Selection

`iloc` selects rows and columns using integer positions.

```python
df.iloc[0]        # First row
df.iloc[-1]       # Last row
df.iloc[0, 1]     # First row, second column
df.iloc[0:3]      # First 3 rows
df.iloc[:, 0]     # First column
df.iloc[:, 0:2]   # First 2 columns
df.iloc[0:3, 1:3] # First 3 rows, columns at positions 1 and 2
```

Key differences:

- `loc` uses labels.
- `iloc` uses integer positions.
- Label-based slices with `loc` include the ending label.
- Position-based slices with `iloc` exclude the ending position.

## 6. Filtering Rows

### Comparison Operators

```python
df[df["age"] > 20]
df[df["age"] < 22]
df[df["age"] >= 21]
df[df["age"] <= 22]
df[df["age"] == 21]
df[df["age"] != 21]
```

### Multiple Conditions

```python
# AND: both conditions must be true
df[(df["age"] > 20) & (df["score"] >= 85)]

# OR: at least one condition must be true
df[(df["age"] < 21) | (df["score"] > 90)]

# NOT: reverse the condition
df[~(df["age"] > 21)]
```

Use parentheses around each condition.

### Filter Using isin()

```python
df[df["name"].isin(["Alice", "Bob"])]
```

Select rows where a column matches one of several values.

### Filter Using between()

```python
df[df["score"].between(80, 95)]
```

By default, both boundary values are included.

### Filter Using query()

```python
df.query("age > 20")
df.query("age > 20 and score >= 85")
df.query("age < 21 or score > 90")
```

`query()` provides a readable way to express many filtering conditions.

## 12. Missing Values

Missing data is commonly represented by `NaN`, `None`, or `pd.NA`.

### Detect Missing Values

```python
df.isna()
df.isnull()
df.isna().sum()
df.isna().sum().sum()
df.notna()
```

`isna()` and `isnull()` are equivalent.

### Remove Missing Values

```python
# Remove rows containing any missing values
df.dropna()

# Remove columns containing any missing values
df.dropna(axis=1)

# Remove rows where all values are missing
df.dropna(how="all")

# Require at least 3 non-missing values in each retained row
df.dropna(thresh=3)

# Check only selected columns
df.dropna(subset=["age", "score"])
```

### Fill Missing Values

```python
# Fill all missing values with zero
df.fillna(0)

# Fill one column
df["age"] = df["age"].fillna(0)

# Fill with mean
df["score"] = df["score"].fillna(df["score"].mean())

# Fill with median
df["age"] = df["age"].fillna(df["age"].median())

# Fill with mode
mode_value = df["age"].mode()
if not mode_value.empty:
    df["age"] = df["age"].fillna(mode_value.iloc[0])
```

`mode()` can return multiple values. Using `.iloc[0]` selects the first mode.

### Forward and Backward Fill

```python
df.ffill()  # Use the previous available value
df.bfill()  # Use the next available value
```

### Fill Only a Specific Column

```python
df["gender"] = df["gender"].bfill()
df["age"] = df["age"].fillna(999)
```

Use placeholder values such as `999` only when they make sense for the dataset. Otherwise, they can distort calculations.

Important: `fillna()` and `dropna()` return a result by default. Assign the result back to keep the changes.

## 17. Grouping with groupby()

`groupby()` follows the split-apply-combine approach:

1. Split the data into groups.
2. Apply an operation to each group.
3. Combine the results.

Example:

```python
data = {
    "department": ["CSE", "EEE", "CSE", "EEE", "CSE"],
    "salary": [30000, 35000, 40000, 45000, 50000]
}

df = pd.DataFrame(data)
```

### Common Grouping Operations

```python
# Mean salary per department
df.groupby("department")["salary"].mean()

# Total salary per department
df.groupby("department")["salary"].sum()

# Number of records per department
df.groupby("department")["salary"].count()

# Multiple statistics
df.groupby("department")["salary"].agg(
    ["mean", "min", "max", "count"]
)

# Group by multiple columns
df.groupby(["department", "city"])["salary"].mean()

# Group by and calculate multiple columns
df.groupby("department").agg({
    "salary": ["mean", "max", "min"]
})
```

`count()` counts non-missing values. `size()` counts rows in each group, including rows where the selected column contains missing values.

```python
df.groupby("department").size()
```

## 19. Combining DataFrames

### concat(): Combine Along an Axis

```python
# Stack rows
combined = pd.concat([df1, df2], axis=0, ignore_index=True)

# Combine columns side by side
combined = pd.concat([df1, df2], axis=1)
```

- `axis=0`: combine rows vertically.
- `axis=1`: combine columns horizontally.
- `ignore_index=True`: create a new sequential index when combining rows.

### merge(): Combine Using Matching Keys

```python
# Only matching keys
pd.merge(df1, df2, on="id", how="inner")

# All left keys
pd.merge(df1, df2, on="id", how="left")

# All right keys
pd.merge(df1, df2, on="id", how="right")

# All keys from both DataFrames
pd.merge(df1, df2, on="id", how="outer")
```

Join types:

- `inner`: only matching keys.
- `left`: all rows from the left DataFrame and matching rows from the right.
- `right`: all rows from the right DataFrame and matching rows from the left.
- `outer`: all keys from both DataFrames.

### Merge When Key Names Differ

```python
pd.merge(
    df1,
    df2,
    left_on="student_id",
    right_on="id",
    how="left"
)
```

### join(): Combine Using Index Labels

```python
df1.join(df2, how="left")
```

`join()` commonly matches DataFrame indexes, while `merge()` commonly matches columns.

- `pivot_table()` can aggregate repeated combinations.
- `melt()` converts multiple columns into rows.

| `df.replace()` | Replace specific values |
| `df.astype()` | Change data types |
| `df.convert_dtypes()` | Infer nullable data types |
| `df.rename()` | Rename columns or index labels |
| `df.reset_index()` | Reset the index |
| `df.sort_values()` | Sort by column values |
| `df.clip()` | Limit values to a specified range |

Example:

```python
df = df.drop_duplicates()
df = df.reset_index(drop=True)
```

## 24. Selecting Data with at and iat

Use these accessors when selecting a single value.

```python
# Label-based scalar access
df.at[0, "name"]

# Position-based scalar access
df.iat[0, 1]

# Set a single value by label
df.at[0, "name"] = "Alex"

# Set a single value by position
df.iat[0, 1] = 25
```

- `at` uses row and column labels.
- `iat` uses integer positions.

## 25. DataFrame vs Series

| Feature | Series | DataFrame |
|---|---|---|
| Dimensions | One-dimensional | Two-dimensional |
| Structure | One labeled sequence | Rows and columns |
| Example | `df["age"]` | `df[["age"]]` |
| Shape | `(rows,)` | `(rows, columns)` |
| Data types | One primary dtype | Different dtypes per column |

Remember:

```python
df["age"]       # Series
df[["age"]]     # DataFrame
```

## 26. Important Concepts to Remember

### Axis

For many DataFrame operations:

- `axis=0`: operate down rows, usually producing a result for each column.
- `axis=1`: operate across columns, usually producing a result for each row.

For example:

```python
df.sum(axis=0)  # Sum each column
df.sum(axis=1)  # Sum across columns for each row
```

### Index

A DataFrame index identifies rows. It does not have to be a sequential integer.

```python
df.index
df.set_index("name")
df.reset_index(drop=True)
```

### In-Place Changes vs Assignment

Many Pandas methods return a new DataFrame instead of changing the existing one.

```python
# Original df is unchanged
df.sort_values("score")

# Keep the sorted result
df = df.sort_values("score")
```

Some methods offer an `inplace` parameter, but assignment is often clearer and easier to maintain.

### Vectorized Operations

Prefer operations on entire columns instead of manually looping through rows.

```python
# Preferred
df["score"] = df["score"] + 5

# Usually unnecessary for simple arithmetic
# for i in range(len(df)):
#     df.loc[i, "score"] += 5
```
