```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Simple Line : y = 2x +1

```python
def fun(x):
  return 2*x+1
```

```python
x = np.linspace(-3,3,100)
y = fun(x)
print(x[:5])
print(y[:5])
```

```python
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Simple Line : 2x+1 ')
plt.grid(True)
plt.show()
print('.'*50)
x1,x2 = -2,1
y1,y2 = fun(x1),fun(x2)
slope = (y2-y1)/(x2-x1)
print(f'Slope : {slope}')
```

Curve : y=x^2

```python
def curve(x):
  return x**2

x = np.linspace(-3,3,100)
y = curve(x)
print(x[:5])
print(y[:5])
plt.plot(x,y)
```

slope = dy/dx
dy = change in y
dx = change in x

```python
def slopeChange(func,x1,h=.00001):
  x2 = x1+h
  y1 = func(x1)
  y2 =func(x2)
  dy = y2-y1
  dx = h
  return dy/dx
```

```python
point = [-2,0,2]
for p in point:
  print(f"{p} -> slope : {slopeChange(curve,p):.2f}")
```

**Weight Update** <br>
w = w - learningRate*slope <br>


>dL/dw > 0, increase loss , so we reduce w
meaning : w = w - lr * (+ve)

> dL/dw < 0, decrease loss , so we increase w
meaning : w = w - lr * (-ve)
* Negative slope tells us to increase the weight.
* Positive slope tells us to decrease the weight.
* Zero slope means we've reached the minimum.







```python
#sample dataset

x = np.array([1,2,3])
y_actual = np.array([2,4,6])
#initial Weight
w = 1.0
#Learning rate
lr = .01

ws,ls = [],[]
# y  = wx predicted
def loss(w):
  y_predicted = w*x
  return np.mean((y_predicted - y_actual)**2)
def dLoss(w):
  y_predicted = w*x
  return np.mean(2*(y_predicted-y_actual)*x)

for i in range(50):
  ws.append(w)
  ls.append(loss(w))
  slope = dLoss(w)
  w = w - lr*slope
  print(f"step: {i+1}, weight: {round(w,3)}, loss: {round(loss(w),3)},slope: {slope}")

ws_plot = np.linspace(0,4,5000)
ls_plot = [loss(wp) for wp in ws_plot]
plt.plot(ws_plot,ls_plot,label='Loss Curve')
plt.scatter(ws,ls,color='red',label='Weight Update')
plt.plot(ws,ls,color='red')
plt.xlabel('Weight')
plt.ylabel('Loss')
plt.legend()
plt.show()
```
