```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Simple Line plot

```python
x = np.linspace(0,10,50,endpoint=True)
#print(x[:5])
y = np.sin(x)
#print(y[:5])
plt.plot(x,y)
plt.title("Simple Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

**Customized Plot**

```python
plt.figure(figsize=(8,6),dpi=80)
plt.plot(x,y,color='red',linestyle='--',linewidth=1,marker='o',label='sin(x)')
plt.title("Customized Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend()
plt.grid(True)
plt.show()
```
