```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Simple Line plot

```python
x = np.linspace(0,10,50,endpoint=True)
#print(x[:5])
y = np.sin(x)
#print(y[:5])
plt.plot(x,y)
plt.title("Simple Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

**Customized Plot**

```python
plt.figure(figsize=(8,6),dpi=80)
plt.plot(x,y,color='red',linestyle='--',linewidth=1,marker='o',label='sin(x)')
plt.title("Customized Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend()
plt.grid(True)
plt.show()
```

**Bar Chart**

```python
students = ['A','B','C','D','E']
scores = [85,90,78,88,60]
plt.bar(students,scores)
plt.title("Bar Chart")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.show()
```

```python
students = ['A','B','C','D','E']
scores = [85,90,78,88,60]
plt.barh(students,scores)
plt.title("Bar Chart")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.show()
```

**Scatter Plot**

```python
rng = np.random.default_rng()
x = rng.random(50)
y = rng.random(50)
plt.scatter(x,y,color='purple')
plt.title("Scatter Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.show()
```

**Histogram**

```python
data = np.random.randn(1000)
plt.hist(data,bins=30,color='green',edgecolor='black')
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

**Pie Chart**

```python
lab = ['A','B','C','D']
sizes = [45,30,15,10]
colors = ['red','green','blue','yellow']
plt.pie(sizes,labels=lab,autopct='%1.1f%%')
plt.show()
```

Multiple Plots(Sub plot)

```python
x = np.linspace(0,10,100)
y = np.sin(x)
plt.figure(figsize=(10,6))
plt.subplot(1,2,1)
plt.plot(x,y,'r--')
plt.title("Sin(x)")
plt.subplot(1,2,2)
plt.plot(x,np.cos(y),'b--')
plt.title("Cos(x)")
plt.show()
```
