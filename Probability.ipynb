```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
rng = np.random.default_rng()
```

#Head or Tail

```python
sample = ['H','T']
n = 500
choice = rng.choice(sample,size=n)
print(choice[:5])
head = np.sum(choice=='H')
tail = np.sum(choice=='T')
print(head/n,np.mean(choice=='H'))
print(f"Percentage : {head/n *100} , {tail/n *100} ")
```

#Dice Roll Simulation

```python
n = 500
roll = rng.integers(1,7,size = n)
values,counts = np.unique(roll,return_counts=True)
print(values,counts)
sum =0
for i in range(len(values)):
  sum +=counts[i]/n
  print(f"Probability of {values[i]} : {(counts[i]/n * 100):.2f}%")
print(sum)
```

```python
#sns.barplot(x=values,y=counts/n)
plt.bar(values,counts/n)
plt.show()
```

#Conditional Probability

100 students 60 like ML of that 60 40 students like DL

```python
df = pd.DataFrame(
    {
        'likesML' : rng.choice([1,0],size=100,p=[0.6,0.4])

    }
)
df['likeDL'] = [
    np.random.choice([1,0],p=[.7,.3]) if ml else
    np.random.choice([1,0],p=[.4,.6])
    for ml in df['likesML']
]
```

#Probability Distribution

```python
n =500
#uniform
plt.hist(rng.uniform(0,1,n),bins=20,alpha=.7,edgecolor='black',label='Uniform')
plt.legend()
plt.show()
#normal
plt.hist(rng.normal(0,1,n),bins=20,alpha=.7,edgecolor='black',label='Normal')
plt.legend()
plt.show()
#bournoulli
plt.hist(rng.binomial(1,.3,n),bins=20,alpha=.7,edgecolor='black',label='Bernoulli')
plt.legend()
plt.show()
#Binomial
plt.hist(rng.binomial(5,.5,n),bins=20,alpha=.7,edgecolor='black',label='Binomial')
plt.legend()
plt.show()
```
