# Conditional Statements & Loops
### Conditional Statement (`if`, `elif`, `else`)

```python
age = int(input("Enter Your Age : "))

if age >= 0 and age <= 18:
  print("Minor")
elif age > 18 and age <= 26:
  print("Adult")
elif age > 26 and age <= 35:
  print("Middle age")
else:
  print("Senior Citizen")
```

```python
numbers = list(map(int,input().split()))
if numbers:
  print("List is not empty")
  num = int(input())
  if num in numbers:
    print(f"{num} is present in list")
  else: print(f"{num} is not present")
else: print("List is empty")
```

```python
student = {
    'name' : 'Tomal',
    'id' : 'B210101017'
}
keyword = input()
if keyword in student:
  print(f"{student['name']}'s student Id is {student['id']}")
else: print("keyword not present")
```

### Loops (`for`, `range`, `while`)

```python
for i in range(5):
  print(i)
```

```python
print(*range(5))
print(*range(0,5))
```

**`range(start, stop, step)`**
* `start` is included, `stop` is excluded.

```python
print(*range(5,10))
print(*range(0,10,2))
print(*range(1,10,2))
```

### Iterating Through Collections (List, Tuple, Set, Dictionary)

```python
arr = [1,2,3,4,5]
tp = (9,8,1,1,4,5)
st = {2,3,4,1,2,2}
mp = {
    'name' : 'tomal',
    'id' : 'B210101017',
    'age' : 24
}
string = 'hello world!'
```

```python
for i in arr:
  print(f"list -> {i}")
print()
for i in st:
  print(f"set -> {i}")
print()
for i in tp:
  print(f"tuple -> {i}")
print("Dictionaries")
for key in mp:
  print(key)
print()
for key in mp.keys():
  print(key)
print()
for key,value in mp.items():
  print(f"key -> {key} , value -> {value}")
```

### While Loop

```python
i = 1
while i <= 5:
  if(i == 2): pass
  else: print(i)
  i += 1
```
