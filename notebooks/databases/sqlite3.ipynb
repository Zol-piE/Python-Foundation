# SQLite3 in Python: Complete Cheat Sheet

## 1. Import SQLite3

```python
import sqlite3
```

## 2. Connect to a Database

```python
conn = sqlite3.connect("school.db")
cursor = conn.cursor()
```

- Creates the database if it does not exist.
- Opens the database if it already exists.
- `conn` represents the database connection.
- `cursor` executes SQL commands.

## 3. Create a Table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER,
    marks REAL
)
""")

conn.commit()
```

### Common SQLite Data Types

| Type | Description |
|---|---|
| `INTEGER` | Whole numbers |
| `REAL` | Decimal numbers |
| `TEXT` | Text or strings |
| `BLOB` | Binary data |
| `NULL` | Missing or null value |

## 4. Insert Data

### Insert One Row

```python
cursor.execute(
    "INSERT INTO students (name, age, marks) VALUES (?, ?, ?)",
    ("Rahim", 21, 85.5)
)

conn.commit()
```

### Insert Multiple Rows

```python
data = [
    ("Karim", 22, 90.0),
    ("Ayesha", 20, 95.5),
    ("Nila", 21, 88.0)
]

cursor.executemany(
    "INSERT INTO students (name, age, marks) VALUES (?, ?, ?)",
    data
)

conn.commit()
```

## 5. Read Data: SELECT

```python
# All columns and rows
cursor.execute("SELECT * FROM students")
print(cursor.fetchall())

# Specific columns
cursor.execute("SELECT name, marks FROM students")
print(cursor.fetchall())

# One row
cursor.execute("SELECT * FROM students")
print(cursor.fetchone())

# First three rows
cursor.execute("SELECT * FROM students LIMIT 3")
print(cursor.fetchall())
```

## 6. Fetch Methods

```python
cursor.fetchone()    # Fetch one row
cursor.fetchmany(3)  # Fetch up to 3 rows
cursor.fetchall()    # Fetch all remaining rows
```

## 7. Filter Rows: WHERE

```python
# Equal to
cursor.execute("SELECT * FROM students WHERE age = 21")

# Greater than
cursor.execute("SELECT * FROM students WHERE marks > 80")

# Less than
cursor.execute("SELECT * FROM students WHERE marks < 90")

# Not equal to
cursor.execute("SELECT * FROM students WHERE age != 21")

# Multiple conditions
cursor.execute(
    "SELECT * FROM students WHERE age > 20 AND marks > 85"
)

# Either condition
cursor.execute(
    "SELECT * FROM students WHERE age = 20 OR marks > 90"
)

# Values in a list
cursor.execute("SELECT * FROM students WHERE age IN (20, 21, 22)")

# Range
cursor.execute("SELECT * FROM students WHERE marks BETWEEN 80 AND 95")

# Missing values
cursor.execute("SELECT * FROM students WHERE marks IS NULL")

# Non-missing values
cursor.execute("SELECT * FROM students WHERE marks IS NOT NULL")
```

## 8. Sort and Limit

```python
# Ascending order
cursor.execute("SELECT * FROM students ORDER BY marks ASC")

# Descending order
cursor.execute("SELECT * FROM students ORDER BY marks DESC")

# Sort by multiple columns
cursor.execute(
    "SELECT * FROM students ORDER BY age ASC, marks DESC"
)

# Limit results
cursor.execute("SELECT * FROM students LIMIT 5")

# Skip the first 5 rows
cursor.execute("SELECT * FROM students LIMIT 5 OFFSET 5")
```

## 9. Search Text: LIKE

```python
# Names starting with A
cursor.execute("SELECT * FROM students WHERE name LIKE 'A%'")

# Names ending with m
cursor.execute("SELECT * FROM students WHERE name LIKE '%m'")

# Names containing ay
cursor.execute("SELECT * FROM students WHERE name LIKE '%ay%'")

# Single-character wildcard
cursor.execute("SELECT * FROM students WHERE name LIKE '_a%'")
```

- `%` matches zero or more characters.
- `_` matches exactly one character.

## 10. Update Data

```python
cursor.execute(
    "UPDATE students SET marks = ? WHERE name = ?",
    (98.0, "Rahim")
)

conn.commit()
```

Update multiple columns:

```python
cursor.execute(
    "UPDATE students SET age = ?, marks = ? WHERE id = ?",
    (22, 92.0, 1)
)

conn.commit()
```

## 11. Delete Data

```python
# Delete a specific row
cursor.execute("DELETE FROM students WHERE id = ?", (1,))
conn.commit()

# Delete all rows
# cursor.execute("DELETE FROM students")
# conn.commit()
```

**Warning:** Always check the `WHERE` condition before running `UPDATE` or `DELETE`.

## 12. Aggregate Functions

```python
# Number of rows
cursor.execute("SELECT COUNT(*) FROM students")

# Average marks
cursor.execute("SELECT AVG(marks) FROM students")

# Maximum marks
cursor.execute("SELECT MAX(marks) FROM students")

# Minimum marks
cursor.execute("SELECT MIN(marks) FROM students")

# Total marks
cursor.execute("SELECT SUM(marks) FROM students")

print(cursor.fetchone()[0])
```

## 13. GROUP BY and HAVING

```python
# Count students by age
cursor.execute("""
SELECT age, COUNT(*)
FROM students
GROUP BY age
""")

print(cursor.fetchall())
```

```python
# Groups with average marks above 85
cursor.execute("""
SELECT age, AVG(marks)
FROM students
GROUP BY age
HAVING AVG(marks) > 85
""")

print(cursor.fetchall())
```

- `GROUP BY` groups rows.
- `HAVING` filters groups.

## 14. DISTINCT: Unique Values

```python
# Unique ages
cursor.execute("SELECT DISTINCT age FROM students")

# Count unique ages
cursor.execute("SELECT COUNT(DISTINCT age) FROM students")
```

## 15. SQL Operators

| Operator | Meaning |
|---|---|
| `=` | Equal |
| `!=` or `<>` | Not equal |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal |
| `<=` | Less than or equal |
| `AND` | Both conditions |
| `OR` | Either condition |
| `NOT` | Negation |
| `IN` | Matches a list |
| `BETWEEN` | Within a range |
| `LIKE` | Pattern matching |
| `IS NULL` | Value is null |

## 16. Parameterized Queries

```python
name = "Rahim"

cursor.execute(
    "SELECT * FROM students WHERE name = ?",
    (name,)
)

print(cursor.fetchall())
```

For multiple parameters:

```python
cursor.execute(
    "SELECT * FROM students WHERE age > ? AND marks > ?",
    (20, 85)
)
```

**Always use placeholders (`?`) for values.** This helps prevent SQL injection and handles values safely.

## 17. Add, Rename, or Remove Columns

```python
# Add a column
cursor.execute("ALTER TABLE students ADD COLUMN city TEXT")

# Rename a table
cursor.execute("ALTER TABLE students RENAME TO learners")

# Rename it back
cursor.execute("ALTER TABLE learners RENAME TO students")

conn.commit()
```

SQLite does not support every form of `ALTER TABLE` in every version. Removing a column is supported in modern SQLite versions.

## 18. Drop a Table

```python
cursor.execute("DROP TABLE IF EXISTS students")
conn.commit()
```

**Warning:** `DROP TABLE` removes the table and its data.

## 19. Primary Key and Constraints

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    email TEXT UNIQUE,
    age INTEGER CHECK (age >= 18),
    department TEXT DEFAULT 'General'
)
""")

conn.commit()
```

- `PRIMARY KEY`: Uniquely identifies a row.
- `AUTOINCREMENT`: Automatically increases generated integer IDs.
- `NOT NULL`: Requires a value.
- `UNIQUE`: Prevents duplicate values.
- `CHECK`: Enforces a condition.
- `DEFAULT`: Supplies a default value.

## 20. Relationships and JOIN

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS departments (
    id INTEGER PRIMARY KEY,
    department_name TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    department_id INTEGER,
    FOREIGN KEY (department_id) REFERENCES departments(id)
)
""")
```

### INNER JOIN

```python
cursor.execute("""
SELECT employees.name, departments.department_name
FROM employees
INNER JOIN departments
ON employees.department_id = departments.id
""")

print(cursor.fetchall())
```

### LEFT JOIN

```python
cursor.execute("""
SELECT employees.name, departments.department_name
FROM employees
LEFT JOIN departments
ON employees.department_id = departments.id
""")
```

**Note:** Enable foreign-key enforcement on a connection with `PRAGMA foreign_keys = ON`.

## 21. Transactions

```python
# Save changes
conn.commit()

# Undo uncommitted changes
conn.rollback()
```

### Context Manager

```python
with sqlite3.connect("school.db") as conn:
    conn.execute(
        "INSERT INTO students (name, age, marks) VALUES (?, ?, ?)",
        ("Sadia", 22, 91.0)
    )
```

The context manager commits on success or rolls back on error. It does not automatically close the connection.

## 22. Read SQLite Data with Pandas

```python
import pandas as pd

df = pd.read_sql_query("SELECT * FROM students", conn)

print(df)
```

### Save a DataFrame to SQLite

```python
df.to_sql("student_data", conn, if_exists="replace", index=False)
```

`if_exists` options:

- `"fail"`: Raise an error if the table exists.
- `"replace"`: Replace the existing table.
- `"append"`: Add rows to the existing table.

## 23. Inspect Database Tables

```python
# List tables
cursor.execute("""
SELECT name FROM sqlite_master
WHERE type = 'table'
""")
print(cursor.fetchall())

# Inspect columns
cursor.execute("PRAGMA table_info(students)")
print(cursor.fetchall())

cursor.execute("SELECT COUNT(*) FROM students")
print(cursor.fetchone()[0])
```

## 24. Database File and In-Memory Database

```python
# Persistent database file
conn = sqlite3.connect("school.db")

# Temporary in-memory database
memory_conn = sqlite3.connect(":memory:")
```

## 25. Close the Connection

```python
cursor.close()
conn.close()
```

## 26. Complete Example

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("school.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER,
    marks REAL
)
""")

data = [
    ("Rahim", 21, 85.5),
    ("Ayesha", 20, 95.5),
    ("Karim", 22, 90.0)
]

cursor.executemany(
    "INSERT INTO students (name, age, marks) VALUES (?, ?, ?)",
    data
)

conn.commit()

cursor.execute("SELECT * FROM students ORDER BY marks DESC")
print(cursor.fetchall())

df = pd.read_sql_query("SELECT * FROM students", conn)
print(df)

conn.close()
```

**Important:** Run the insert section repeatedly only when you intend to add duplicate records. A new database connection does not clear an existing database.
