# Functions & Object-Oriented Programming (Classes)

```python
def fun():
  print("Function one")
fun()
```

```python
def add(a,b):
  return a+b
print(add(10,20))
```

```python
def func(name,age):
  print(name,age)
func('tomal',20)          # Positional
func(age=80,name='Tonu')  # Keyword Argument
```

### Parameter Ordering
`def fun(positional, /, normal, *args, keyword_only, **kwargs):`

```python
def fun(a,b,*args,**kargs):
  print(a,b,args,kargs)
fun(1,2,3,4,5,6,name='tomal',age=12)
```

```python
def fun(a,b,/,*,c,d):
  print(a,b,c,d)
fun(10,20,c=30,d=40)
```

### Global Scope, Decorators, and Type Hints

```python
name = 'tomal'
age = 10
def fun1():
  global name,age
  name = 'Sowa'
  age = 8
print(name,age)
fun1()
print(name,age)
```

```python
def decorator(func):
  def wrap():
    print("before")
    func()
    print("After")
  return wrap

@decorator
def fun():
  print("Function")

fun()
```

```python
# Specific return type using -> type
def add(a,b) -> int:
  return a+b
def div(a,b) -> float:
  return a/b
print(add(10,20))
print(div(10,3))
```

### Classes in Python

```python
class student:
  name = 'tomal'
  age = 20

s1 = student()
print(s1.name,s1.age)
```

```python
class student:
  school = "CSTU" # Class Attribute
  def __init__(self,name,age):
    self.name = name # Object Attribute
    self.age = age

s1 = student('tomal',20)
print(s1.name,s1.age,s1.school)
s2 = student('Tonu',21)
print(s2.name,s2.age,s2.school)

s1.school = "NSTU"
print(s1.name,s1.age,s1.school)
print(s2.name,s2.age,s2.school)
student.school = "DU"
print(s1.name,s1.age,s1.school)
print(s2.name,s2.age,s2.school)
s2.gender = 'Female'
print(s2.name,s2.age,s2.school,s2.gender)
```

```python
class student:
  school = "CSTU" # Class Attribute
  def __init__(self,name,age):
    self.name = name # Object Attribute
    self.age = age
  @classmethod
  def change_school(cls,name):
    cls.school = name
  @staticmethod
  def info():
    print("This is student class")

s1 = student('tomal','20')
s1.gender = 'male'
print(s1.name,s1.age,s1.school,s1.gender)
student.change_school('NSTU')
print(s1.name,s1.age,s1.school,s1.gender)
s2 = student('Tonu','21')
print(s2.name,s2.age,s2.school)
student.info()
```
