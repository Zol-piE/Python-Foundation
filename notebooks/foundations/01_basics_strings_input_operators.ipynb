# Strings, User Input, and Operators
### String and Function

```python
name = 'tomal islam tohidul'
print(name[2:8])
print(name.title())
print(name.replace('tomal','tohidul'))
```

### Basic List Indexing

```python
arr = [1,2,3,4,5]
nameNumber = ['tomal','tonu',12,'islam']
print(arr[2:])
print(nameNumber[-1])
```

### User Input
* User input always comes in string

```python
num = input('Enter a Integer number ')
print(f"User enter number : {num}")
print(type(num))
num = int(num)
print(type(num))
```

```python
a,b = input('enter two number ').split()
print(a,b)
a,b,c = input('Enter three numbers ').split()
print(a,b,c)
```

```python
# convert multiple inputs into int
a,b = map(int,input("Enter Two Numbers ").split())
print(type(a))
print(type(b))
print(a,b)
```

```python
# split divides the string into list
text = "tomal nabil rohomot"
names = text.split()
print(names)
```

```python
# map(function,collection)
numbers = list(map(int,['1','2','3']))
print(numbers)
id = list(map(str,[2,17,27]))
print(id)
```

### Membership Operator
Test membership in a sequence:
* **in** : True if value present
* **not in** : True if value not present

```python
numbers = [1,2,3,4,5,6]
num = int(input())
print(num in numbers)
```

### Identity Operator
Compare memory location of two objects:
* **is** : True if same object
* **is not** : True if different object

```python
a = [1,2,3]
b = [1,2,3]
c = a
print(a is b)
print(c is a)
print(a is not b)
```
