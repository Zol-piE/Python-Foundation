# Python Data Structures: Sets & Tuples
A **set** is a collection of unique elements in Python:
- Unordered
- Mutable
- Does not allow duplicate values
- Does not support indexing
- Can store different data types
- Elements must be immutable/hashable (integers, strings, tuples)

```python
a = {1,2,3,4,5}
b = {10,20,30,40}
a.add(11)
print(a)
a.update([23,34,45])
print(a)
# remove() -> Error if not present
# discard() -> does not give error if not present
a.remove(3)
# a.remove(55)
print(a)
a.discard(55)
print(a)
```

### Set Operations

```python
a = {1,2,3,4,5}
b = {4,5,10,20}
# union
print(a.union(b))
print(a|b)
# intersection
print(a.intersection(b))
print(a&b)
# difference
print(a.difference(b))
print(a-b)
# Symmetric Difference
print(a.symmetric_difference(b))
print(a^b)
```

```python
a = {1,2,3,4,5,6}
b = {4,5}
print(b.issubset(a))
print(a.issuperset(b))
print(a.isdisjoint(b))
```
