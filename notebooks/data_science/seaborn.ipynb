```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
print(sns.__version__)
```

**Load Dataset**

```python
df = sns.load_dataset('tips')
print(df.info())
print(df.shape)
print(df.describe())
```

```python
sns.set_style('darkgrid')
sns.set_palette('viridis')
```

**Histogram - Distrubution of total bill**

```python
plt.figure(figsize=(10,5))
sns.histplot(df['total_bill'], bins=20,kde=True)
plt.title('Histogram of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

```python
df.head()
```

**Box Plot**

```python
plt.figure(figsize=(10,9))
sns.boxplot(x='day',y='total_bill',data=df,hue='day',legend=False)
plt.title('Box Plot of Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

**Count Plot - Number of records per Day**

```python
sns.countplot(x='day',data=df,hue='sex')
plt.title('Count Plot of Number of Records per Day')
plt.xlabel('Day')
plt.ylabel('Frequency')
plt.show()
```

**Scatter Plot**

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x='total_bill',y='tip',data=df,hue='sex',size='size')
plt.title('Scatter Plot of Total Bill vs Tip')
```

Pair Plot - Pair Wise Feature Relationship

> `sns.pairplot()` is a Seaborn function that visualizes relationships between multiple numerical variables in a dataset at once.



```python
sns.pairplot(df,hue='sex')
```

# **Heatmap**


> a heatmap helps you quickly identify which numerical features are strongly related to one another.<br>


Correlation	Interpretation
* +1.00	Perfect positive linear relationship
* +0.80	Strong positive linear relationship
* 0.00	No linear relationship
* -0.80	Strong negative linear relationship
* -1.00	Perfect negative linear relationship



```python
# Select numerical features
ndf = df.select_dtypes(include='number')
dfcorr = ndf.corr()
sns.heatmap(dfcorr,annot=True,cmap='coolwarm',vmin=-1,vmax=1)

```
