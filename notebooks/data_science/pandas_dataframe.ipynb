# **Pandas DataFrame** <br>
> A DataFrame in Pandas is a two-dimensional, tabular data structure with labeled rows and columns. It is similar to an Excel spreadsheet or a SQL table.



```python
data = {
    'name' : ['Alice', 'Bob', 'Claire'],
    'age' : [25, 30, 27],
    'gender' : ['Female', 'Male', 'Female']
}
print(data)
print(type(data))
```

```python
import pandas as pd
df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
print(df.head())
print('________')
print(df.tail())
```

```python
df = pd.read_csv('dataSample.csv')
print(df)
new_df = pd.DataFrame({
    'name' : ['tomal', 'tonu', 'sowa'],
    'age' : [25, 30, 27],
    'gender' : ['Male', 'Female', 'Female']
})
print(new_df,'\n')
df = pd.concat([df,new_df],ignore_index=True)
df['score'] = [60,35,30,75,55,30]
print(df)
```

```python
new_df = [{
    'name':'tahmid','age':25,'score':75
},
{'name':'rifat','age':22,'gender':'Female',},
{'name': 'sayed','gender':'Male','score':80}
  ]
df = pd.concat([df,pd.DataFrame(new_df)],ignore_index=True)
print(df)
df.to_csv('dataSample2.csv',index=False)
```

# **removing elements**

```python
#Removing Rows
new_df = df.drop(index=6).reset_index(drop=True)
#print(new_df)
new_df = df.drop(df.index[1])
#print(new_df)
```

```python
#Removing Columns
new_df = df.drop(columns=['score'])
#print(new_df)
new_df = df.drop('score',axis=1)
#print(new_df)
new_df = df[df['score']>=80]
print(new_df)
```

```python
new_df = df.dropna()                    # Remove rows containing missing values
print(new_df)
new_df =df.dropna(axis=1)              # Remove columns containing missing values
new_df =df.dropna(subset=['age'])      # Remove rows with missing age
new_df =df.dropna(how='all')           # Remove rows where all values are missing
```

# **Handling Null Values**

```python
new_df = df.isna()
#print(new_df)
new_df = df.notna()
#print(new_df)
new_df = df.isna().sum()
print(new_df)
```

```python
#Fill Missing Values
new_df = df.fillna(0)
#print(new_df)
new_df = df.fillna(method='ffill')
#print(new_df)
new_df = df.fillna(method='bfill')
#print(new_df)
df['age'] = df['age'].fillna(999)
df['gender'] = df['gender'].bfill()
print(df['score'].mode())
df['score'] =df['score'].fillna(df['score'].mode()[0])
print(df)
```
