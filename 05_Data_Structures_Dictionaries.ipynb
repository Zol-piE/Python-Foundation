### Dictionary Cheatsheet

```python
d = {"a": 1, "b": 2}

d["a"]                 # Access
d.get("a")             # Safe access
d["c"] = 3             # Add
d["a"] = 10            # Update
del d["a"]             # Delete
d.pop("b")             # Remove and return
d.popitem()            # Remove last pair
d.keys()               # All keys
d.values()             # All values
d.items()              # All pairs
d.update({"c": 3})     # Update multiple items
d.setdefault("x", 0)   # Insert default if absent
d.copy()               # Shallow copy
d.clear()              # Empty dictionary
len(d)                 # Number of items
"x" in d               # Check key
```

### Dictionary Comprehension & Merging

```python
d1 = {x: x+2 for x in range(1,5)}
print(d1)
d2 = {x: x+2 for x in range(5,10)}
print(d2)
merge = d1 | d2 # merge two dictionaries
print(merge)
```

### Nested Dictionaries

```python
students = {
    'std1' : {
        'name' : 'tomal',
        'id' : 'B210101017',
        'age' : 24
    },
    'std2' : {
        'name' : 'tom',
        'id' : 'B210101018',
        'age' : 25
    },
}
print(students)
print(students['std1']['name'])
```
