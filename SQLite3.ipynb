

> ***SQL (Structured Query Language)*** is a language used to store, retrieve, update, and delete data in relational databases.


> ***SQLite*** is a lightweight, file-based relational database that uses SQL to manage data. It doesn't require a separate database server.use for embedded databse system






```python
import sqlite3
```

```python
#connect to a SQLite database
connection = sqlite3.connect('example.db')
```

```python
cursor = connection.cursor()
```

```python
#create a table
cursor.execute(
    '''
    Create Table if Not exists employees(
    id integer primary key,
    name text not null,
    age integer,
    department text
    )
    '''
)
#commit changes
connection.commit()
```

```python
cursor.execute(
  '''
  select * from employees
  '''
)
```

```python
cursor.execute(
    '''
    insert into employees (name, age, department)
    values ('John Doe', 30, 'IT')
    '''
)
cursor.execute(
    '''
    insert into employees (name, age, department)
    values ('tomal', 30, 'SWE')
    '''
)
cursor.execute(
    '''
    insert into employees (name, age, department)
    values ('Minhaj', 50, 'Marketing')
    '''
)
connection.commit()
```

```python
#Query From the Table
cursor.execute('select * from employees')
rows = cursor.fetchall()
#print Query data
for row in rows:
  print(row)
```

```python
#update data in table
cursor.execute(
    '''
    update employees
    set  age=18
    where name='tomal'
    '''
)
connection.commit()
```

```python
#Query From the Table
cursor.execute('select * from employees')
rows = cursor.fetchall()
#print Query data
for row in rows:
  print(row)
```

```python
#delete the data
cursor.execute(
    '''
    delete from employees
    where name='John Doe'
    '''
)
connection.commit()
```

```python
#Query From the Table
cursor.execute('select * from employees')
rows = cursor.fetchall()
#print Query data
for row in rows:
  print(row)
```

```python
connection = sqlite3.connect('salesData.db')
cursor = connection.cursor()
cursor.execute(
    '''
    create table if not exists sales(
    id integer primary key,
    date text not null,
    product text not null,
    sales integer,
    region text
    )
    '''
)
connection.commit()
```

```python
sales_data = [
    ('2022-01-01', 'Product A', 100, 'North'),
    ('2022-01-02', 'Product B', 150, 'South'),
    ('2022-01-03', 'Product C', 200, 'East'),
    ('2022-01-04', 'Product A', 120, 'West'),
]
```

```python
cursor.executemany(
    '''
    insert into sales(date,product,sales,region)
      values(?,?,?,?)
    ''',
    sales_data
)
connection.commit()
```

```python
cursor.execute('select * from sales')
rows = cursor.fetchall()
#print Query data
for row in rows:
  print(row)
```
