# NumPy Random Sampling Operations

## 1. Random Sampling Operations

```python
import numpy as np

rng = np.random.default_rng(42)
arr = np.array([10, 20, 30, 40, 50])
```

| Method | Description | Syntax |
|---|---|---|
| `random()` | Generates random floats from 0 to 1 | `rng.random(5)` |
| `choice()` | Selects random elements from an array | `rng.choice(arr, size=3)` |
| `shuffle()` | Shuffles the original array in place | `rng.shuffle(arr)` |
| `permutation()` | Returns a randomly rearranged array | `rng.permutation(arr)` |
| `bytes()` | Generates random bytes | `rng.bytes(10)` |

### Examples

```python
# Random floats
rng.random(5)

# Select random elements
rng.choice(arr, size=3)

# Select without replacement (no duplicates)
rng.choice(arr, size=3, replace=False)

# Shuffle original array
rng.shuffle(arr)

# Return shuffled copy
rng.permutation(arr)

# Generate 10 random bytes
rng.bytes(10)
```

## 2. Quick Reference: Random Distributions

| Goal | Method |
|---|---|
| Random floats between 2 and 8 | `rng.uniform(2, 8, size=5)` |
| Random integers from 1 to 9 | `rng.integers(1, 10, size=5)` |
| Values around an average | `rng.normal(50, 10, size=5)` |
| Random proportions between 0 and 1 | `rng.beta(2, 5, size=5)` |
| Number of successes in fixed trials | `rng.binomial(10, 0.5, size=5)` |
| Counts of events | `rng.poisson(4, size=5)` |
| Waiting times | `rng.exponential(2, size=5)` |
| Randomly select existing values | `rng.choice(arr, size=5)` |
| Generate related variables | `rng.multivariate_normal(mean, cov, size=5)` |

## 3. Important Notes

- `size=5` generates 5 values.
- `size=(2, 3)` generates an array with 2 rows and 3 columns.
- `replace=False` prevents duplicate selections within a sample.
- `shuffle()` modifies the original array.
- `permutation()` returns a rearranged array without modifying the original.
- `rng.uniform()` generates floating-point values.
- `rng.integers()` generates integer values.
- `rng = np.random.default_rng(42)` initializes a reproducible random generator.

**Remember:** A distribution determines how random values are generated, while sampling operations select or rearrange values.

# NumPy Array Combining:

## 1. concatenate() vs stack()

These are the two most important array-combining functions.

- concatenate(): Joins arrays along an existing axis.
- stack(): Combines arrays along a new axis. Input arrays must have the same shape.

```python
import numpy as np

a = np.array([1, 2])
b = np.array([3, 4])

np.concatenate((a, b))
# [1 2 3 4]       shape: (4,)

np.stack((a, b))
# [[1 2]
#  [3 4]]         shape: (2, 2)
```

## 2. vstack() vs hstack() vs dstack()

These functions stack arrays in different directions.

### vstack()

- Stacks arrays vertically.
- For 1D arrays, converts each array into a row.

```python
np.vstack((a, b))
# [[1 2]
#  [3 4]]
```

### hstack()

- For 1D arrays, joins elements into one array.
- For 2D arrays, combines columns horizontally.

```python
np.hstack((a, b))
# [1 2 3 4]
```

### dstack()

- Stacks arrays along the third axis (axis=2 for 3D arrays).
- Useful for combining depth layers or channels.

```python
np.dstack((a, b))
# [[[1 3]
#   [2 4]]]
```

### Relationship with concatenate()

For 2D arrays:
- vstack() is equivalent to concatenate(..., axis=0).
- hstack() is equivalent to concatenate(..., axis=1).

For 1D arrays, hstack() and concatenate() produce the same result, but vstack() produces rows.

## 3. column_stack() vs hstack()

- hstack(): Joins 1D arrays into a single 1D array.
- column_stack(): Converts 1D arrays into columns.

```python
np.hstack((a, b))
# [1 2 3 4]       shape: (4,)

np.column_stack((a, b))
# [[1 3]
#  [2 4]]         shape: (2, 2)
```

For 2D arrays, both combine columns horizontally.

## 4. row_stack() vs vstack()

Both produce the same result. row_stack() is an alias for vstack().

```python
np.row_stack((a, b))
np.vstack((a, b))
```

Prefer vstack() for clarity.

## 5. append() vs concatenate()

Both can join arrays, but their behavior differs.

```python
np.append(a, b)
# [1 2 3 4]

np.concatenate((a, b))
# [1 2 3 4]
```

- append(): Adds values to an array. Without axis, it flattens the inputs.
- concatenate(): Joins arrays along an existing axis.

Prefer concatenate() when combining arrays, especially when working with multidimensional arrays.

## 6. block()

Combines arrays in a specified block layout.

```python
a = np.array([[1, 2]])
b = np.array([[3, 4]])
c = np.array([[5, 6]])
d = np.array([[7, 8]])

np.block([
    [a, b],
    [c, d]
])

# [[1 2 3 4]
#  [5 6 7 8]]
```

Use block() when you want to arrange arrays in a grid.

## 7. np.r_ vs np.c_

These are convenient shortcuts for joining arrays.

```python
np.r_[a, b]  # Join along the first axis
np.c_[a, b]  # Combine as columns
```

- np.r_: Similar to concatenation for 1D arrays.
- np.c_: Similar to column_stack() for 1D arrays.

## 8. Quick Comparison

| Function | Main Purpose |
|---|---|
| concatenate() | Join along an existing axis |
| stack() | Join along a new axis |
| vstack() | Stack vertically |
| hstack() | Stack horizontally |
| dstack() | Stack along the third axis |
| column_stack() | Convert 1D arrays into columns |
| row_stack() | Same as vstack() |
| append() | Append values to an array |
| block() | Arrange arrays in a block layout |
| np.r_ | Convenient array joining |
| np.c_ | Convenient column combining |

1. concatenate(): General-purpose array joining.
2. stack(): Combining arrays along a new dimension.
3. vstack(): Adding rows.
4. hstack(): Joining horizontally.
5. column_stack(): Converting 1D arrays into columns.
