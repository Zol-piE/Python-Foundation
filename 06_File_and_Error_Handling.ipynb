# File Handling & Error Handling
* `open` statement needs closing (`f.close()`)
* `with` statement requires no manual closing

**Modes:**
* `"r"` : Read a file
* `"w"` : Write, overwrite existing content
* `"a"` : Append content to the end
* `"x"` : Create a new file, error if it exists
* `"rb"` : Read binary file
* `"wb"` : Write binary file
* `"r+"` : Read and write

```python
f = open('newTextFile.txt','w')
f.write("New line 1") # Erase Everything and write this
f.close()
```

```python
f = open('newTextFile.txt','a')
f.write("adds new words with old one")
f.write("\nadds new Line with old one")
f.close()
```

```python
f = open('newTextFile.txt','r')
for line in f:
  print(line)
f.close()
```

### Using the `with` Statement

```python
lines = ['line1\n','line2\n','line3\n']
with open('newTextFile.txt','w') as f:
  f.writelines(lines)
```

```python
with open('newTextFile.txt','a') as f:
  f.writelines('This is a New Line')
```

```python
with open('newTextFile.txt','r') as f:
  for line in f:
    print(line)
```

### File Pointer Navigation
* **`seek(i)`** -> moves the pointer to $i$-th position
* **`tell()`** -> gets the current position of pointer in file

```python
with open('newTextFile.txt','r') as f:
  content = f.read(8)
  print(f.tell())
  f.seek(3)
  print(f.read())
```

### Error Handling (`try`, `except`, `finally`)

```python
try:
  num = int(input())
except Exception as e:
  print(f"Exception Occured {str(e)}")
finally:
  print("Executes whatever Happens")
```

```python
try:
  num = int(input())
  x = 10/num
  print(x)
except ValueError:
  print("Enter A number")
except ZeroDivisionError:
  print("Cant be Divided by Zero")
```
